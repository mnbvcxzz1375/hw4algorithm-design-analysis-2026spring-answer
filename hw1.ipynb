## 问题 1

根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$

**解答：**

**1. $T(n)=T(n/4)+1$**

迭代展开：

$T(n)=T(n/4)+1=T(n/4^2)+2=\cdots=T(n/4^k)+k$

当 $n/4^k=1$ 即 $k=\log_4 n$ 时终止，$T(n)=T(1)+\log_4 n=1+\log_4 n$。

因此 $T(n)=\Theta(\log n)$。

---

**2. $T(n)=3T(n/3)+n^2$**

运用主定理。$a=3$，$b=3$，$f(n)=n^2$。

$\log_b a=\log_3 3=1$，$f(n)=n^2=\Omega(n^{1+\epsilon})$，其中取 $\epsilon=1>0$，属于主定理情况3。

检查正则条件：$af(n/b)=3\cdot (n/3)^2=n^2/3\le cn^2$，取 $c=1/3<1$，满足。

故 $T(n)=\Theta(n^2)$。

---

**3. $T(n)=T(2n/3)+1$**

子问题规模为 $2n/3$，$b=3/2$，直接迭代展开：

$T(n)=T\left(\frac{2}{3}n\right)+1=T\left(\left(\frac{2}{3}\right)^2 n\right)+2=\cdots=T\left(\left(\frac{2}{3}\right)^k n\right)+k$

令 $(2/3)^k n=1$，解得 $k=\log_{3/2} n=\frac{\log n}{\log(3/2)}$。

因此 $T(n)=\Theta(\log n)$。

---

**4. $T(n)=5T(n/4)+n$**

$a=5$，$b=4$，$\log_b a=\log_4 5\approx 1.16$。

$f(n)=n=O(n^{\log_4 5-\epsilon})$，其中取 $\epsilon=\log_4 5-1\approx 0.16>0$，属于主定理情况1。

故 $T(n)=\Theta(n^{\log_4 5})$。

---

**5. $T(n)=T(\sqrt{n})+1$，$n>2$；$T(n)=1$，$n\le 2$**

令 $n=2^m$，定义 $S(m)=T(2^m)$。

由 $\sqrt{n}=2^{m/2}$，代入原式：

$T(2^m)=T(2^{m/2})+1\quad\Rightarrow\quad S(m)=S(m/2)+1$

其中边界：$n\le 2$ 即 $2^m\le 2$，$m\le 1$，$S(1)=1$。

该递归形式与第1问相同，$S(m)=\Theta(\log m)$。

由 $m=\log n$，得 $T(n)=S(\log n)=\Theta(\log\log n)$。

## 问题 2

观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)
-----
n := r-p+1
if n=2 and A[p]>A[r] then
    swap A[p] and A[r]
else if n >= 3 then
    m = ceil(2n/3)
    DoSomething(A,p,p+m-1)
    DoSomething(A,r-m+1,r)
    DoSomething(A,p,p+m-1)  
    
---
first call: DoSomething(A,1,n)
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案
3. \* 使用python编程实现上述算法或其对应的更高效版本的算法

idea:

**思路分析：**

**1. 时间复杂度递归公式：**

- 当 $n=2$ 时：只做一次比较和可能的交换，$T(2)=O(1)$
- 当 $n\ge 3$ 时：调用三次递归，每次处理 $\lceil 2n/3 \rceil$ 个元素

递归公式：$T(n)=3T(\lceil 2n/3 \rceil)+O(1)$

使用主定理近似分析：$a=3, b=3/2$，$\log_{3/2} 3 = \log 3 / \log(3/2) \approx 2.71$

$f(n)=O(1)=O(n^{\log_{3/2}3 - \epsilon})$，属于情况1

因此 $T(n)=O(n^{\log_{3/2} 3}) \approx O(n^{2.71})$

**2. 算法功能：**

分析递归过程：
- 处理前 $\lceil 2n/3 \rceil$ 个元素
- 处理后 $\lceil 2n/3 \rceil$ 个元素（与前部分有重叠）
- 再次处理前 $\lceil 2n/3 \rceil$ 个元素

该算法实现的是**排序功能**（类似冒泡排序的分治版本）。

但效率很低：$O(n^{2.71})$ 远高于最优排序算法 $O(n\log n)$。

**更高效的方案：** 使用快速排序、归并排序等，时间复杂度 $O(n\log n)$。

````python
# 问题2：实现分治排序算法及其高效版本

import math

def do_something(A, p, r):
    """原分治排序算法"""
    n = r - p + 1
    if n == 2 and A[p] > A[r]:
        A[p], A[r] = A[r], A[p]  # 交换
    elif n >= 3:
        m = math.ceil(2 * n / 3)
        do_something(A, p, p + m - 1)
        do_something(A, r - m + 1, r)
        do_something(A, p, p + m - 1)

def quick_sort(arr):
    """高效排序算法：快速排序"""
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

# 测试
A = [5, 3, 8, 1, 2, 7, 4, 6]
A_copy = A.copy()
do_something(A, 0, len(A) - 1)  # 原算法（索引从0开始）
print(f"原算法 DoSomething 结果: {A}")
print(f"高效算法 快速排序 结果: {quick_sort(A_copy)}")
print("算法时间复杂度: O(n^2.71) ≈ O(n^{log_{3/2} 3})")
print("高效版本时间复杂度: O(n log n)")
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

**思路分析：**

**理解问题：** profits[i] 是第i天的收益（涨跌），需要计算从买入日到卖出日之间的累计收益。题目要求卖出日在买入日之后，因此至少要选择一个长度不少于2天的合法区间。

例如：第5天买入，第8天卖出，收益 = profits[5] + profits[6] + profits[7] + profits[8] = 5 + 2 + (-1) + 3 = 9

**转化为前缀和问题：**

定义前缀和 prefix[i] = profits[1] + profits[2] + ... + profits[i]

则从第a天买入，第b天卖出的收益 = prefix[b] - prefix[a-1]

**O(n)算法设计：**

1. 计算前缀和数组 prefix
2. 问题转化为：找 prefix[b] - prefix[a-1] 的最大值，其中 b > a
3. 这等价于：对于每个位置b，找 prefix[0..b-2] 中最小的 prefix[a-1]
4. 遍历过程中维护当前最小前缀和，计算最大差值

如果所有区间收益都是负数，也应返回亏损最少的合法区间；若题目允许"不交易"，再把答案与0取最大即可。

**时间复杂度：O(n)**，只需一次遍历

**空间优化：** 无需显式构建前缀和数组，在遍历过程中用一个变量维护当前前缀和即可，**空间复杂度降为 O(1)**。

````python
# 问题3：股票最大利润问题 O(n) 解法（O(1)空间优化）

def max_profit(profits):
    """
    计算股票最大利润
    使用前缀和方法，一次遍历解决问题，无需额外数组，空间复杂度 O(1)
    返回: (最大收益, 买入日期, 卖出日期)，日期从1开始计数，要求卖出日 > 买入日
    """
    n = len(profits)
    if n < 2:
        return None, None, None

    max_gain = float('-inf')
    buy_day = sell_day = None
    min_prefix_val = 0       # prefix[0] = 0
    min_prefix_idx = 0
    current_prefix = 0       # 当前前缀和

    for i in range(n):
        current_prefix += profits[i]
        if i >= 1:  # 至少隔一天才能卖出
            gain = current_prefix - min_prefix_val
            if gain > max_gain:
                max_gain = gain
                buy_day = min_prefix_idx + 1
                sell_day = i + 1
        # 更新最小前缀和（供后续天数作为买入候选）
        if current_prefix < min_prefix_val:
            min_prefix_val = current_prefix
            min_prefix_idx = i + 1

    return max_gain, buy_day, sell_day

# 测试
profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
gain, buy, sell = max_profit(profits)
print(f"测试 profits={profits}")
print(f"最大收益: {gain}")
print(f"买入日期: 第{buy}天")
print(f"卖出日期: 第{sell}天")
print("算法时间复杂度: O(n), 空间复杂度: O(1)")
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为"局部最小"。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:

**思路分析：**

**1. O(log n) 找一个局部最小（二分搜索）：**

关键观察：由于边界条件 A[1]>A[2] 和 A[n-1]<A[n]，数组两端都"向下"，因此必定存在局部最小。

算法设计（类似二分搜索）：
1. 检查中间元素 A[mid]
2. 如果 A[mid] 是局部最小，返回
3. 否则，A[mid] 必有一侧邻居比它小（假设左边小），则左侧一定存在局部最小
4. 递归搜索有更小邻居的那一侧

每次将搜索范围减半，时间复杂度 O(log n)

**2. O(n) 找所有局部最小：**

遍历数组，检查每个元素是否满足局部最小条件。时间复杂度 O(n)。

````python
# 问题4：局部最小值问题

def find_one_local_min(arr):
    """
    O(log n) 使用二分搜索找到一个局部最小值
    题目中的局部最小值必须同时拥有左右邻居，因此只搜索下标1到n-2。
    """
    n = len(arr)
    if n < 3:
        return None

    left, right = 1, n - 2
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] < arr[mid - 1] and arr[mid] < arr[mid + 1]:
            return arr[mid]
        # 向较小邻居一侧搜索，该侧必定存在局部最小
        if arr[mid] > arr[mid + 1]:
            left = mid + 1
        else:
            right = mid - 1
    return None

def find_all_local_min(arr):
    """
    O(n) 遍历找到所有局部最小值
    """
    result = []
    n = len(arr)
    for i in range(1, n - 1):
        if arr[i] < arr[i - 1] and arr[i] < arr[i + 1]:
            result.append(arr[i])
    return result

# 测试
A = [9, 3, 7, 2, 1, 4, 5]
print(f"测试数组: {A}")
print(f"O(log n) 找到一个局部最小: {find_one_local_min(A)}")
print(f"O(n) 找到所有局部最小: {find_all_local_min(A)}")
print("算法时间复杂度: O(log n) 和 O(n)")
````

## 问题 5

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

**思路分析：**

**1. 排序方法：**

选择快速排序（或归并排序），时间复杂度 O(n log n)，然后取前k个。

总时间复杂度：O(n log n)

**2. 堆方法（建堆 + 提取）：**

- 建堆：使用 `heapify` 原地建堆，时间复杂度 O(n)，优于逐个 `heappush` 的 O(n log n)
- Extract-min：每次 O(log n)，执行k次，O(k log n)

总时间复杂度：O(n + k log n)

**3. 大数据处理方法（维护最大堆）：**

维护一个大小为k的最大堆（存储当前最小的k个数）：
- 读入新数据，如果比堆顶小，替换堆顶并调整
- 每次操作 O(log k)
- 处理n个数据，总时间 O(n log k)

最后将堆中元素排序输出：O(k log k)

总时间复杂度：O(n log k)

````python
# 问题5：找k个最小数

import heapq

def k_min_sort(arr, k):
    """
    方法1：排序后取前k个
    时间复杂度: O(n log n)
    """
    k = min(k, len(arr))
    sorted_arr = sorted(arr)
    return sorted_arr[:k]

def k_min_heap_insert(arr, k):
    """
    方法2：建堆后提取k次
    使用 heapify 建堆，时间复杂度 O(n)，优于逐个 heappush 的 O(n log n)
    提取k次时间复杂度: O(k log n)
    总时间复杂度: O(n + k log n)
    """
    k = min(k, len(arr))
    heap = arr.copy()
    heapq.heapify(heap)  # O(n) 原地建堆，替代 n 次 heappush
    result = []
    for _ in range(k):
        result.append(heapq.heappop(heap))  # O(log n)
    return result

def k_min_large_data(arr, k):
    """
    方法3：大数据处理，维护大小为k的最大堆
    使用最大堆存储当前最小的k个数（堆顶为这k个数中最大的）
    时间复杂度: O(n log k)
    """
    if k <= 0:
        return []
    # Python heapq是最小堆，用负数模拟最大堆
    max_heap = []  # 存储负数，模拟最大堆
    for x in arr:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -x)  # 堆未满，直接加入
        elif x < -max_heap[0]:  # 新数比堆顶（最大）小，替换
            heapq.heapreplace(max_heap, -x)  # O(log k)
    # 提取并排序结果
    result = sorted([-x for x in max_heap])
    return result

# 测试
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3
print(f"测试数组: {arr}")
print(f"方法1 (排序): 前{k}个最小数为 {k_min_sort(arr, k)}, 时间复杂度 O(n log n)")
print(f"方法2 (堆): 前{k}个最小数为 {k_min_heap_insert(arr, k)}, 时间复杂度 O(n + k log n)")
print(f"方法3 (大数据): 前{k}个最小数为 {k_min_large_data(arr, k)}, 时间复杂度 O(n log k)")
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:

**思路分析：**

**关键观察：**

- 正数星球向右运动，负数星球向左运动
- 只有**正数在前、负数在后**的情况才会发生碰撞（方向相反）
- 负数在前、正数在后：方向相同（都向右），不会碰撞
- 负数在前、负数在后：都向左，不会碰撞
- 正数在前、正数在后：都向右，不会碰撞

**使用栈模拟：**

1. 遍历数组，用栈存储存续的星球
2. 对于负数星球：可能与栈顶正数星球碰撞
3. 碰撞规则：质量大的存活；质量相等时，负数消失（题目规定）
4. 继续碰撞直到栈顶不是向右星球、当前星球消失，或栈空

**时间复杂度：O(n)**，每个元素最多入栈出栈一次

````python
# 问题6：星球碰撞问题

def planet_collision(A):
    """
    使用栈模拟星球碰撞
    时间复杂度: O(n)
    """
    stack = []
    for planet in A:
        alive = True
        while alive and stack and stack[-1] > 0 and planet < 0:
            if abs(stack[-1]) > abs(planet):
                alive = False
            elif abs(stack[-1]) < abs(planet):
                stack.pop()
            else:
                # 质量相等时，自右向左运动的星球消失，自左向右运动的星球保留
                alive = False
        if alive:
            stack.append(planet)
    return stack

# 测试
A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
A2 = [23, -8, 9, -3, -7, 9, -23, 22]
print(f"测试1: A={A1}, 输出: {planet_collision(A1)}")
print(f"测试2: A={A2}, 输出: {planet_collision(A2)}")
print("算法时间复杂度: O(n)")
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

**思路分析：**

**0-1背包问题：** 每个物品只能选择装入或不装入。

**动态规划解法：**

定义 dp[i][w]：考虑前i个物品，背包容量为w时的最大价值

状态转移方程：
- 如果不选第i个物品：dp[i][w] = dp[i-1][w]
- 如果选第i个物品（重量为wi）：dp[i][w] = dp[i-1][w-wi] + vi
- dp[i][w] = max(dp[i-1][w], dp[i-1][w-wi] + vi)

**时间复杂度：O(nW)**，其中n是物品数量，W是背包容量

**空间优化：** 可以用一维数组，O(W)空间

````python
# 问题7：0-1背包问题

def knapsack_01(items, capacity):
    """
    0-1背包问题动态规划解法
    items: [(重量, 价值)] 列表
    capacity: 背包最大容量
    返回: 最大价值
    时间复杂度: O(n*W)
    """
    n = len(items)
    # dp[w] 表示容量为w时的最大价值
    dp = [0] * (capacity + 1)
    # 记录选择方案
    selected = [[] for _ in range(capacity + 1)]
    
    for i, (weight, value) in enumerate(items):
        for w in range(capacity, weight - 1, -1):  # 逆序遍历，避免重复使用
            if dp[w - weight] + value > dp[w]:
                dp[w] = dp[w - weight] + value
                selected[w] = selected[w - weight] + [i + 1]  # 记录选择的物品编号
    
    return dp[capacity], selected[capacity]

# 测试
items = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9
max_value, selected_items = knapsack_01(items, capacity)
print(f"大理石: {items}")
print(f"背包容量: {capacity}")
print(f"最大价值: {max_value}")
print(f"选择的物品: 大理石{selected_items}")
print(f"算法时间复杂度: O(nW) = O({len(items)}*{capacity}) = O({len(items)*capacity})")
````

## 问题 8

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g.:
---
1.

<div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-2.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

**思路分析：**

**关键观察：** 二叉搜索树的中序遍历应该是严格递增的。

如果两个节点被交换，中序遍历序列中会出现两个"反序"：
- 第一个反序：前一个节点值 > 后一个节点值
- 如果是相邻节点交换，只有一处反序
- 如果是不相邻节点交换，有两处反序

**算法设计：**

1. 中序遍历BST，记录遍历序列
2. 找出序列中的反序节点（第一个反序的前节点，最后一个反序的后节点）
3. 交换这两个节点的值

**时间复杂度：O(n)**，一次遍历

**空间优化：** 可以在遍历时直接记录反序节点，无需存储完整序列。

````python
# 问题8：恢复二叉搜索树

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(values):
    """从列表构建二叉树（层序）"""
    if not values:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root

def inorder(root):
    """中序遍历"""
    result = []
    stack = []
    curr = root
    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        result.append(curr.val)
        curr = curr.right
    return result

def recover_bst(root):
    """
    恢复二叉搜索树
    时间复杂度: O(n)
    """
    # 中序遍历找到反序节点
    first, second = None, None
    prev = None
    stack = []
    curr = root
    
    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        # 检查反序
        if prev and prev.val > curr.val:
            if not first:
                first = prev  # 第一个反序的前节点
            second = curr  # 最后一个反序的后节点
        prev = curr
        curr = curr.right
    
    # 交换两个节点的值
    if first and second:
        first.val, second.val = second.val, first.val

# 测试
root1 = build_tree([1, 3, None, None, 2])
print(f"测试1: root=[1,3,null,null,2]")
recover_bst(root1)
print(f"恢复后的中序遍历: {inorder(root1)}")

root2 = build_tree([3, 1, 4, None, None, 2])
print(f"测试2: root=[3,1,4,null,null,2]")
recover_bst(root2)
print(f"恢复后的中序遍历: {inorder(root2)}")
print("算法时间复杂度: O(n)")
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。



<div style="text-align:center;width:800px;"> <img alt="" src="./fig/hw3q6.png"> </div>

请分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

**思路分析：**

**关键观察：** 每个位置能接的雨水取决于其左右两侧最高柱子的较小值。

位置i能接的水量 = min(左边最高, 右边最高) - height[i]

**O(n)双指针解法：**

1. 使用左右双指针，分别从两端向中间移动
2. 维护左端最高 left_max 和右端最高 right_max
3. 比较 left_max 和 right_max：
   - 如果 left_max < right_max：左边位置的水量由 left_max 决定，左指针右移
   - 否则：右边位置的水量由 right_max 决定，右指针左移

**时间复杂度：O(n)**，一次遍历

**空间复杂度：O(1)**，只使用常数空间

````python
# 问题9：接雨水问题

def trap_rain_water(height):
    """
    使用双指针计算接雨水量
    时间复杂度: O(n)
    空间复杂度: O(1)
    """
    if not height:
        return 0
    left, right = 0, len(height) - 1
    left_max, right_max = height[left], height[right]
    water = 0
    
    while left < right:
        if left_max < right_max:
            left += 1
            left_max = max(left_max, height[left])
            water += left_max - height[left]
        else:
            right -= 1
            right_max = max(right_max, height[right])
            water += right_max - height[right]
    return water

# 测试
height1 = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
height2 = [4, 2, 0, 3, 2, 5]
print(f"测试1: height={height1}, 雨水量: {trap_rain_water(height1)}")
print(f"测试2: height={height2}, 雨水量: {trap_rain_water(height2)}")
print("算法时间复杂度: O(n)")
print("空间复杂度: O(1)")
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:

**思路分析：**

**DFS遍历：** 从根节点遍历到每个叶子节点，累计路径上的数字。

**算法设计：**

1. DFS遍历，传递当前累计数字值
2. 到达叶子节点时，累加到总和
3. 数字计算：new_sum = old_sum * 10 + node.val

**时间复杂度：O(n)**，遍历所有节点

**空间复杂度：O(h)**，递归栈深度，h为树高度

````python
# 问题10：二叉树路径数字之和

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(values):
    """从列表构建二叉树（层序）"""
    if not values:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root

def sum_numbers(root):
    """
    DFS遍历计算所有路径数字之和
    时间复杂度: O(n)
    空间复杂度: O(h)
    """
    def dfs(node, current_sum):
        if not node:
            return 0
        current_sum = current_sum * 10 + node.val
        # 叶子节点：返回当前数字
        if not node.left and not node.right:
            return current_sum
        # 非叶子节点：继续递归
        return dfs(node.left, current_sum) + dfs(node.right, current_sum)
    
    return dfs(root, 0)

# 测试
root1 = build_tree([1, 2, 3])
print(f"测试1: root=[1,2,3], 路径数字之和: {sum_numbers(root1)}")

root2 = build_tree([4, 9, 0, 5, 1])
print(f"测试2: root=[4,9,0,5,1], 路径数字之和: {sum_numbers(root2)}")
print("算法时间复杂度: O(n)")
print("空间复杂度: O(h)")
````
